```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("df_clean2.csv")
```

```python
df['group'].value_counts()
```

<h2> Treatment vs control group distribution

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="group")
plt.title("Treatment vs control group distribution")
plt.xlabel("Students group")
plt.ylabel("Number of students")
plt.show()
```

<h2> number of students at different courses

```python
df['course'].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, 
              x='course', 
             order=df['course'].value_counts().index
)
plt.title("course-wise students distribution")
plt.xlabel("courses")
plt.ylabel("Number of students")
plt.show()
```

<h2> online vs offline students 

```python
df['mode'].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data = df , x='mode')
plt.title("online vs offline students")
plt.xlabel("Mode")
plt.ylabel("Number of students")
plt.show()
```

<h2> city-wise student distribution

```python
df['city'].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, 
              x='city', 
              order=df['city'].value_counts().index, 
              hue='city'
)
plt.title("city vs student distribution")
plt.xlabel('cities')
plt.ylabel("Number of students")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, 
              y='city', 
              order=df['city'].value_counts().index, 
              hue='city'
)
plt.title("city vs student distribution")
plt.ylabel('cities')
plt.xlabel("Number of students")
plt.show()
```

<h2> Average final score by group

```python
plt.figure(figsize=(6,4))
sns.barplot(data=df , x='group', y='final_score')
plt.title("average final score by group")
plt.show()
```

<h2> distribution of pre score

```python
plt.figure(figsize=(6,4))
sns.histplot(
    data=df, 
    x='pre_score', 
    bins=20, 
    kde=True 
)
plt.title("distribution of pre score")
plt.xlabel("pre-score")
plt.ylabel("number of students")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(
    data=df, 
    x='final_score', 
    bins=20, 
    kde=True 
)
plt.title("distribution of final_score")
plt.xlabel("final-score")
plt.ylabel("number of students")
plt.show()
```

<h2> outlier detection 

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=df , x = 'pre_score')
plt.title("boxplot of pre-score")
plt.xlabel("pre-score")
plt.show()
```

<h2>Attendance vs Final Score – Relationship Analysis

```python
plt.figure(figsize=(8, 5)) 
sns.scatterplot(
    data= df, 
    x="attendance_pct",
    y="final_score",
    alpha=0.5
)
plt.title("Attendance vs Final Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Final Score")

plt.show()
```

<h2>Correlation Analysis

```python
correlation = df[
    ["pre_score", "final_score",
     "attendance_pct", "score_improvement"
    ]
].corr()
correlation
```

```python
plt.figure(figsize=(8, 5)) 
sns.heatmap( correlation , annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
