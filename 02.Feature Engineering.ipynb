Feature Engineering

```python
df['score_improvement'] = (
    df['final_score']-df['pre_score']
)
```

```python
df[["student_id",
    "pre_score",
    "final_score",
    "score_improvement"]].head(10)
```

```python
df['performance_change'] = np.select(
    [
        df['score_improvement'] > 0 , 
        df['score_improvement'] < 0
    ], 
    [
        "Improved", 
        "Decline" 
    ], 
    default="No Change"
    
)
```

```python
df['performance_change'].value_counts()
```

```python
df.loc[
    df['performance_change'].isna(), 
    "performance_change"
] = 'Unknown'
```

```python
df['performance_change'].value_counts()
```

```python
df['enrolment_month'] = (
    df['enrolment_date'].dt.month
)
```

```python
df['enrolment_date'].dtype
```

```python
df['enrolment_date'] = pd.to_datetime(
    df['enrolment_date'],
    errors='coerce'
)
```

```python
df['enrolment_month'] = (
    df['enrolment_date'].dt.month
)
```

```python
df['enrolment_date'].dtype
```

```python
df['enrolment_day'] = (
    df['enrolment_date'].dt.day
)
```

```python
df['month_name'] = (
    df['enrolment_date'].dt.month_name()
)
```
