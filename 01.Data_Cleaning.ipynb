```python
df_clean = df.copy()
```

```python
df_clean["attendance_pct"] = pd.to_numeric(
    df_clean["attendance_pct"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)
```

```python
df_clean["pre_score"] = df_clean["pre_score"].fillna(
    df_clean["pre_score"].mean()
)

df_clean["attendance_pct"] = df_clean["attendance_pct"].fillna(
    df_clean["attendance_pct"].median()
)
```

```python
df_clean.isnull().sum()
```

```python
df_clean.duplicated().sum()
```

```python
df_clean[df_clean.duplicated()]
```

```python
# Remove duplicate student records
df_clean = df_clean.drop_duplicates(
    subset="student_id",
    keep="first"
)
```

```python
df_clean.shape
```

```python
df_clean.duplicated().sum()
```

```python
#cleaning cities data 
df_clean['city'].unique()
```

```python
df_clean["city"] = (
    df_clean["city"]
    .str.strip()
    .str.lower()
    .str.title()
    .replace({
        "Vadodra": "Vadodara",
        "Gandhinagr": "Gandhinagar"
    })
)
```

```python
df_clean['city'].unique()
```

```python
#cleaning mode column
df_clean['mode'] = ( 
    df_clean['mode']
    .str.strip()
    .str.lower()
    .str.title()
)
```

```python
df_clean['mode'].unique()
```

```python
#finding invalid score 
invalid_score = df_clean[
    (df_clean['final_score'] < 0) | 
    (df_clean['final_score'] > 100) 
]
print("invalid score:",len(invalid_score))
```

```python
df_clean.loc[
    (df_clean['final_score'] < 0) | 
    (df_clean['final_score'] > 100 ), 
    "final_score"
] = np.nan
```

```python
df_clean["final_score"].describe()
```

```python
invalid_score = df_clean[
    (df_clean['final_score'] < 0) | 
    (df_clean['final_score'] > 100) 
]
print("invalid score:",len(invalid_score))
```

```python
invalid_attendence = df_clean[
    (df_clean['attendance_pct'] < 0) | 
   ( df_clean['attendance_pct'] > 100)
]
print("invalid attendence: ",len(invalid_attendence))
```

```python
df_clean.loc[
    (df_clean["attendance_pct"] < 0 ) |
    (df_clean["attendance_pct"] > 100), 
    "attendance_pct"
] = np.nan
```

```python
invalid_attendence = df_clean[
    (df_clean['attendance_pct'] < 0) | 
   ( df_clean['attendance_pct'] > 100)
]
print("invalid attendence: ",len(invalid_attendence))
```

```python
# dates cleaning 
df_clean["enrolment_date"] = pd.to_datetime(
    df_clean["enrolment_date"], 
    format="mixed", 
    dayfirst= True
)
```
